```python
import urllib.request
import urllib.error
import json
import pandas as pd
import time
import os
from datetime import datetime

# Configuration
RESOURCE_ID = "eb40ac75-8976-5bdc-bbe8-434fa3a0b0ae"
API_URL = f"https://discover.data.vic.gov.au/api/3/action/datastore_search?resource_id={RESOURCE_ID}&limit=10000"
OUTPUT_FILE = "../data/raw/live_harvester_august_2026.csv"

print("Deploying Autonomous API Harvester (v2: Fault-Tolerant)...")
print(f"Target Output: {OUTPUT_FILE}")
print("Press the 'Interrupt Kernel' (Stop) button in Jupyter to safely halt the loop on Saturday.\n")

pull_count = 0

try:
    while True:
        try:
            # 1. Fetch live data
            req = urllib.request.urlopen(API_URL)
            api_json = json.loads(req.read().decode('utf-8'))
            records = api_json["result"]["records"]
            
            if records:
                df = pd.DataFrame(records)
                
                # 2. Inject execution timestamp
                df["harvest_timestamp"] = datetime.now().isoformat()
                
                # 3. Append to CSV
                if not os.path.isfile(OUTPUT_FILE):
                    df.to_csv(OUTPUT_FILE, index=False)
                else:
                    df.to_csv(OUTPUT_FILE, mode='a', header=False, index=False)
                    
                pull_count += 1
                print(f"[{datetime.now().strftime('%H:%M:%S')}] Pull #{pull_count} verified. Appended {len(df)} records.")
            else:
                print(f"[{datetime.now().strftime('%H:%M:%S')}] API returned empty payload.")
                
        except urllib.error.HTTPError as e:
            print(f"[{datetime.now().strftime('%H:%M:%S')}] Server Error ({e.code}). Data.vic is down. Retrying next cycle.")
        except urllib.error.URLError as e:
            print(f"[{datetime.now().strftime('%H:%M:%S')}] Network Error ({e.reason}). Check your connection. Retrying next cycle.")
        except Exception as e:
            print(f"[{datetime.now().strftime('%H:%M:%S')}] Unexpected local error: {e}. Retrying next cycle.")
            
        # 4. Sleep for 5 minutes (300 seconds) unconditionally
        print("Sleeping 5 mins...")
        time.sleep(300)

except KeyboardInterrupt:
    print("\n--- Harvester Terminated by User ---")
    print(f"Total successful pulls: {pull_count}")
```
